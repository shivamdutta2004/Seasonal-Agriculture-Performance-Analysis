The dataset covers three agricultural seasons: Kharif, Rabi and Zaid, along with multiple crops and states. This project uses the dataset to analyze how agricultural performance varies across seasons and to identify meaningful patterns, differences and relationships in the available data.

```python
clean_df = df.copy()

clean_df["Rainfall_mm"] = clean_df.groupby("Season")["Rainfall_mm"].transform(
    lambda x: x.fillna(x.median())
)

clean_df["Soil_Moisture_pct"] = clean_df.groupby("Season")["Soil_Moisture_pct"].transform(
    lambda x: x.fillna(x.median())
)

print("Missing values after filling Rainfall and Soil Moisture:")
print(clean_df[["Rainfall_mm", "Soil_Moisture_pct"]].isnull().sum())
```

### Treatment of Missing Yield Values

The 32 missing values in `Yield_Tonnes_Ha` were not imputed because yield is the primary outcome variable of the analysis. Records with missing yield values were excluded only from analyses that require yield values, while the remaining records were retained for other analyses.

```python
categorical_columns = ["Season", "Crop", "State", "Irrigation_Method"]

for column in categorical_columns:
    print(f"\n{column}:")
    print(clean_df[column].unique())
```

```python
season_count = clean_df["Season"].value_counts()

print("Number of records in each season:")
print(season_count)
```

```python
season_percentage = clean_df["Season"].value_counts(normalize=True) * 100

print("Percentage of records in each season:")
print(season_percentage.round(2))
```

```python
# Q1. How does agricultural performance vary across seasons?
# Supporting analysis: Which characteristics change between seasons?

season_performance = clean_df.groupby("Season").agg({
    "Yield_Tonnes_Ha": ["mean", "median"],
    "Production_Tonnes": ["mean", "median"],
    "Revenue_INR": ["mean", "median"],
    "Total_Cost_INR": ["mean", "median"],
    "Profit_INR": ["mean", "median"],
    "Water_Used_m3": ["mean", "median"],
    "Water_Efficiency_t_per_1000m3": ["mean", "median"],
    "Disease_Pest_Risk_pct": ["mean", "median"]
})

season_performance = season_performance.reindex(["Kharif", "Rabi", "Zaid"])

season_performance.round(2)
```

### Interpretation

Kharif shows the strongest overall agricultural performance, with the highest median yield (1.95 tonnes/ha), median profit (₹38,808), profitable-farm percentage (57.79%), and water efficiency (3.28 tonnes per 1000 m³). Rabi shows intermediate performance, while Zaid has the lowest median yield (1.45 tonnes/ha), lowest median profit (−₹62,144), and lowest percentage of profitable farms (35.52%). These results indicate clear differences in agricultural performance across the three seasons.

```python
# Q2. What major seasonal patterns can be observed?

season_yield = clean_df.groupby("Season")["Yield_Tonnes_Ha"].median()
season_yield = season_yield.reindex(["Kharif", "Rabi", "Zaid"])

plt.figure(figsize=(8, 5))

plt.bar(season_yield.index, season_yield.values)

plt.title("Median Agricultural Yield by Season")
plt.xlabel("Season")
plt.ylabel("Median Yield (Tonnes/Ha)")

plt.tight_layout()
plt.show()
```

##Interpretation

Median agricultural yield decreases consistently from Kharif (1.95 tonnes/ha) to Rabi (1.66 tonnes/ha) and Zaid (1.45 tonnes/ha). This shows a clear declining seasonal yield pattern, with Kharif having the highest median yield and Zaid the lowest.

```python
# Q3. Which season has the strongest economic performance in terms of median profit and percentage of profitable farms?

season_profit = clean_df.groupby("Season")["Profit_INR"].median()
season_profit = season_profit.reindex(["Kharif", "Rabi", "Zaid"])

print("Median Profit by Season:")
print(season_profit)
```

```python
# Q6. What relationships exist between environmental conditions and agricultural yield?

environmental_columns = [
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Humidity_pct",
    "Sunlight_Hours_Day",
    "Soil_pH",
    "Soil_Moisture_pct"
]

environment_yield_corr = clean_df[
    environmental_columns + ["Yield_Tonnes_Ha"]
].corr()["Yield_Tonnes_Ha"].drop("Yield_Tonnes_Ha")

print("Correlation of environmental factors with yield:")
print(environment_yield_corr.sort_values(ascending=False).round(3))
```

```python
# Q7. What relationships exist between resource usage and agricultural yield?

resource_columns = [
    "Water_Used_m3",
    "Fertilizer_kg_ha",
    "Nitrogen_kg_ha",
    "Phosphorus_kg_ha",
    "Potassium_kg_ha",
    "Pesticide_Litre_ha"
]

resource_yield_corr = clean_df[
    resource_columns + ["Yield_Tonnes_Ha"]
].corr()["Yield_Tonnes_Ha"].drop("Yield_Tonnes_Ha")

print("Correlation of resource-usage variables with yield:")
print(resource_yield_corr.sort_values(ascending=False).round(3))
```

```python
from scipy.stats import kruskal
```

```python
# Q10. Are the observed differences in agricultural yield and profit across seasons statistically significant?

kharif_yield = clean_df[
    clean_df["Season"] == "Kharif"
]["Yield_Tonnes_Ha"].dropna()

rabi_yield = clean_df[
    clean_df["Season"] == "Rabi"
]["Yield_Tonnes_Ha"].dropna()

zaid_yield = clean_df[
    clean_df["Season"] == "Zaid"
]["Yield_Tonnes_Ha"].dropna()

statistic, p_value = kruskal(
    kharif_yield,
    rabi_yield,
    zaid_yield
)

print("Kruskal-Wallis H-statistic:", round(statistic, 3))
print("p-value:", p_value)
```

```python
# Supporting analysis: Is profit significantly different across seasons?

kharif_profit = clean_df[
    clean_df["Season"] == "Kharif"
]["Profit_INR"].dropna()

rabi_profit = clean_df[
    clean_df["Season"] == "Rabi"
]["Profit_INR"].dropna()

zaid_profit = clean_df[
    clean_df["Season"] == "Zaid"
]["Profit_INR"].dropna()

profit_statistic, profit_p_value = kruskal(
    kharif_profit,
    rabi_profit,
    zaid_profit
)

print("Kruskal-Wallis H-statistic for Profit:",
      round(profit_statistic, 3))

print("p-value for Profit:",
      profit_p_value)
```

### Interpretation

The Kruskal–Wallis test indicates statistically significant differences across the three seasons for both agricultural yield and profit. For yield, H = 68.604 with p < 0.05, and for profit, H = 101.926 with p < 0.05. Therefore, the results provide statistical evidence that the distributions of yield and profit differ across Kharif, Rabi and Zaid.

## Final Findings

1. Kharif showed the highest median agricultural yield at 1.95 tonnes/ha, followed by Rabi at 1.66 tonnes/ha and Zaid at 1.45 tonnes/ha.

2. Kharif showed the strongest economic performance, with a median profit of ₹38,808 and 57.79% of farms being profitable.

3. Zaid showed the weakest economic performance, with a median profit of -₹62,144 and only 35.52% of farms being profitable.

4. Median yield decreased from Kharif to Rabi to Zaid across all eight crops in the dataset.

5. Drip irrigation had the highest median yield (1.96 tonnes/ha) and median profit (₹52,895), while Flood irrigation had the lowest median yield (1.63 tonnes/ha) and median profit (-₹12,131).

6. Rainfed irrigation had the highest median water efficiency at 4.58 tonnes per 1000 m³, showing that water efficiency and profitability do not necessarily lead to the same ranking.

7. The selected environmental variables showed very weak linear relationships with yield. Rainfall had the highest correlation among them at only 0.028.

8. Water usage had the strongest relationship with yield among the selected resource variables, with a correlation of 0.389. Higher water-use groups also showed higher median yield.

9. Telangana had the highest median yield among the states at 1.84 tonnes/ha, while Andhra Pradesh had the lowest at 1.65 tonnes/ha.

10. The Kruskal-Wallis test indicated statistically significant differences in yield across seasons (H = 68.604, p < 0.05) and in profit across seasons (H = 101.926, p < 0.05).
